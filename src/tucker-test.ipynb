# Transpose

```python
def transpose(a, dim):
    n1, n2, n3 = a.shape
    if dim == 120:
        a = np.reshape(a, (n1, n2*n3))
        a = a.T
        a = np.reshape(a, (n2, n3, n1))
    elif dim == 201:
        a = np.reshape(a, (n1*n2, n3))
        a = a.T
        a = np.reshape(a, (n3, n1, n2))
    elif dim == 210:
        a_tmp = a.copy()
        a = a.flatten()
        a_tmp = a_tmp.flatten()
        for i in range(n1):
            for j in range(n2):
                for k in range(n3):
                    a[k * n2 * n1 + j * n1 + i] = a_tmp[i * n2 * n3 + j * n3 + k]
        a = np.reshape(a, (n3, n2, n1))
    elif dim == 21:
        a_tmp = a.copy()
        a = a.flatten()
        a_tmp = a_tmp.flatten()
        for i in range(n1):
            for j in range(n2):
                for k in range(n3):
                    a[i * n2 * n3 + k * n2 + j] = a_tmp[i * n2 * n3 + j * n3 + k]
        a = np.reshape(a, (n1, n3, n2))
    else:
        print("WRONG dim")
    return a
```

```python
a = np.random.rand(32, 34, 5)

print(np.linalg.norm(np.transpose(a, (1, 2, 0)) - transpose(a, 120)))
print(np.linalg.norm(np.transpose(a, (2, 1, 0)) - transpose(a, 210)))
print(np.linalg.norm(np.transpose(a, (2, 0, 1)) - transpose(a, 201)))
print(np.linalg.norm(np.transpose(a, (0, 2, 1)) - transpose(a,  21)))
```
